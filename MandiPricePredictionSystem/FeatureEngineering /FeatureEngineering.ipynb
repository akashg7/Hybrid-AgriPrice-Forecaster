```python
import numpy as np
```

```python
df["date"] = pd.to_datetime(df["date"])
df = df.sort_values(["Commodity", "Mandi", "date"]).reset_index(drop=True)
```

# **1. TEMPORAL FEATURES**

```python
df["day"] = df["date"].dt.day
df["month"] = df["date"].dt.month
df["year"] = df["date"].dt.year
df["day_of_week"] = df["date"].dt.dayofweek
df["week_of_year"] = df["date"].dt.isocalendar().week.astype(int)
df["day_of_year"] = df["date"].dt.dayofyear
df["is_weekend"] = (df["day_of_week"] >= 5).astype(int)

# Fourier seasonal encoding
df["sin1"] = np.sin(2 * np.pi * df["day_of_year"] / 365)
df["cos1"] = np.cos(2 * np.pi * df["day_of_year"] / 365)
df["sin2"] = np.sin(4 * np.pi * df["day_of_year"] / 365)
df["cos2"] = np.cos(4 * np.pi * df["day_of_year"] / 365)
```

```python
df.head()
```

# **2. PRICE LAG FEATURES**

```python
group = df.groupby(["Commodity", "Mandi"])

for lag in [1, 3, 7, 14, 30]:
    df[f"modal_lag_{lag}"] = group["ModalPrice"].shift(lag)
```

```python
df.head()
```

# **3. PRICE ROLLING FEATURES**

```python
for window in [3, 7, 14, 30]:
    df[f"rolling_mean_{window}"] = group["ModalPrice"].transform(lambda x: x.rolling(window).mean())
    df[f"rolling_std_{window}"] = group["ModalPrice"].transform(lambda x: x.rolling(window).std())
```

```python
df.head(50)
```

# **4. VOLATILITY FEATURES**

```python
df["price_range"] = df["MaxPrice"] - df["MinPrice"]
df["price_spread"] = df["MaxPrice"] - df["ModalPrice"]
df["norm_spread"] = df["price_range"] / (df["ModalPrice"] + 1)

df["volatility_7"] = df["rolling_std_7"]
df["volatility_30"] = df["rolling_std_30"]

df["zscore_7"] = (df["ModalPrice"] - df["rolling_mean_7"]) / (df["rolling_std_7"] + 1e-6)
df["momentum_7"] = df["ModalPrice"] - df["modal_lag_7"]
df["momentum_14"] = df["ModalPrice"] - df["modal_lag_14"]
```

```python
df.head(50)
```

# **5. ARRIVAL FEATURES**

```python
for lag in [1, 3, 7, 14]:
    df[f"arrivals_lag_{lag}"] = group["Arrivals"].shift(lag)

for window in [7, 30]:
    df[f"arrivals_avg_{window}"] = group["Arrivals"].transform(lambda x: x.rolling(window).mean())

df["arrival_change_7"] = df["Arrivals"] - df["arrivals_lag_7"]
```

```python
df.head(50)
```

# **6. WEATHER FEATURES (AND ANOMALIES)**

```python
df["temp_range"] = df["temp_max"] - df["temp_min"]
df["rain_intensity"] = df["rainfall"] / 24

# Rolling 30-day climate trend
df["temp_avg_30"] = group["temp_avg"].transform(lambda x: x.rolling(30).mean())
df["rain_avg_30"] = group["rainfall"].transform(lambda x: x.rolling(30).mean())

df["temp_anomaly"] = df["temp_avg"] - df["temp_avg_30"]
df["rain_anomaly"] = df["rainfall"] - df["rain_avg_30"]
```

```python
df.head(50)
```

# **7. Combined Spatial Features (Raw + Transformed)**

```python
# --- Spatial Features (Combined Raw + Sin/Cos) ---
df["lat"] = df["lat"].astype(float)
df["lon"] = df["lon"].astype(float)

df["lat_sin"] = np.sin(np.radians(df["lat"]))
df["lat_cos"] = np.cos(np.radians(df["lat"]))
df["lon_sin"] = np.sin(np.radians(df["lon"]))
df["lon_cos"] = np.cos(np.radians(df["lon"]))

spatial_features = [
    "lat", "lon",
    "lat_sin", "lat_cos",
    "lon_sin", "lon_cos"
]
```

```python
df = df.sort_values(["Commodity", "Mandi", "date"]).reset_index(drop=True)

df["series_id"], series_mapping = pd.factorize(
    df["Commodity"].astype(str) + " | " + df["Mandi"].astype(str)
)
```

# **2. Define STATIC features**

```python

#Static Categorical (embed as integers)
static_cat_cols = ["Commodity", "CropGroup", "State", "district_name", "Mandi", "Variety", "Grade"]

# Encode to integer IDs
for col in static_cat_cols:
    df[col + "_id"] = df[col].astype("category").cat.codes

#Static Numeric (spatial)
static_num_cols = ["lat_sin", "lat_cos", "lon_sin", "lon_cos"]
```
