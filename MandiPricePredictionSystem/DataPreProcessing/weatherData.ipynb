## Visualize Key Weather Metrics Over Time

### Subtask:
Create visualizations, such as line plots, to show the trends of average temperature, rainfall, and humidity over time across all districts, or for a selected district if trends vary significantly.


**Reasoning**:
I will import the `matplotlib.pyplot` and `seaborn` libraries to prepare for generating the visualizations as requested in the instructions.



```python
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully.")
```

**Reasoning**:
To visualize the trends of average temperature, rainfall, and humidity over time, I will first resample the DataFrame to a monthly frequency and then create separate line plots for each metric.



```python
df_monthly_avg = df.resample('M', on='date')[['temp_avg', 'rainfall', 'humidity']].mean()

plt.figure(figsize=(15, 5))
sns.lineplot(data=df_monthly_avg, x=df_monthly_avg.index, y='temp_avg')
plt.title('Monthly Average Temperature Over Time')
plt.xlabel('Date')
plt.ylabel('Average Temperature (°C)')
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(15, 5))
sns.lineplot(data=df_monthly_avg, x=df_monthly_avg.index, y='rainfall')
plt.title('Monthly Average Rainfall Over Time')
plt.xlabel('Date')
plt.ylabel('Average Rainfall (mm)')
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(15, 5))
sns.lineplot(data=df_monthly_avg, x=df_monthly_avg.index, y='humidity')
plt.title('Monthly Average Humidity Over Time')
plt.xlabel('Date')
plt.ylabel('Average Humidity (%)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

**Reasoning**:
The previous code generated a `FutureWarning` regarding the use of 'M' in `resample`. To resolve this and ensure future compatibility, I will replace 'M' with 'ME' as suggested in the warning.



```python
df_monthly_avg = df.resample('ME', on='date')[['temp_avg', 'rainfall', 'humidity']].mean()

plt.figure(figsize=(15, 5))
sns.lineplot(data=df_monthly_avg, x=df_monthly_avg.index, y='temp_avg')
plt.title('Monthly Average Temperature Over Time')
plt.xlabel('Date')
plt.ylabel('Average Temperature (°C)')
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(15, 5))
sns.lineplot(data=df_monthly_avg, x=df_monthly_avg.index, y='rainfall')
plt.title('Monthly Average Rainfall Over Time')
plt.xlabel('Date')
plt.ylabel('Average Rainfall (mm)')
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(15, 5))
sns.lineplot(data=df_monthly_avg, x=df_monthly_avg.index, y='humidity')
plt.title('Monthly Average Humidity Over Time')
plt.xlabel('Date')
plt.ylabel('Average Humidity (%)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Analyze Weather Parameters by District

### Subtask:
Calculate and visualize the average values of key weather parameters (e.g., 'temp_avg', 'rainfall', 'humidity') for each 'district' to identify regional differences.


**Reasoning**:
To analyze weather parameters by district, I will first group the DataFrame `df` by the 'district' column and calculate the mean for 'temp_avg', 'rainfall', and 'humidity' to get the average values for each district. I'll store the results in `df_district_avg`.



```python
df_district_avg = df.groupby('district')[['temp_avg', 'rainfall', 'humidity']].mean().reset_index()
print("Average weather parameters per district:")
print(df_district_avg.head())
```

**Reasoning**:
To visualize the average weather parameters by district, I will create three separate horizontal bar plots for 'temp_avg', 'rainfall', and 'humidity', sorting each by its respective average value to enhance readability and identify regional differences.



```python
parameters = ['temp_avg', 'rainfall', 'humidity']
labels = {'temp_avg': 'Average Temperature (°C)', 'rainfall': 'Average Rainfall (mm)', 'humidity': 'Average Humidity (%)'}
titles = {'temp_avg': 'Average Temperature by District', 'rainfall': 'Average Rainfall by District', 'humidity': 'Average Humidity by District'}

for param in parameters:
    df_sorted = df_district_avg.sort_values(by=param, ascending=False)

    plt.figure(figsize=(12, df_sorted.shape[0] * 0.2))
    sns.barplot(x=param, y='district', data=df_sorted, palette='viridis')
    plt.title(titles[param])
    plt.xlabel(labels[param])
    plt.ylabel('District')
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```

**Reasoning**:
The previous code generated a `FutureWarning` due to `palette` being used without `hue` in `sns.barplot`. To resolve this and ensure future compatibility, I will explicitly set `hue='district'` and `legend=False` in the `barplot` call as suggested by the warning.



```python
parameters = ['temp_avg', 'rainfall', 'humidity']
labels = {'temp_avg': 'Average Temperature (°C)', 'rainfall': 'Average Rainfall (mm)', 'humidity': 'Average Humidity (%)'}
titles = {'temp_avg': 'Average Temperature by District', 'rainfall': 'Average Rainfall by District', 'humidity': 'Average Humidity by District'}

for param in parameters:
    df_sorted = df_district_avg.sort_values(by=param, ascending=False)

    plt.figure(figsize=(12, df_sorted.shape[0] * 0.2))
    sns.barplot(x=param, y='district', data=df_sorted, palette='viridis', hue='district', legend=False)
    plt.title(titles[param])
    plt.xlabel(labels[param])
    plt.ylabel('District')
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```
