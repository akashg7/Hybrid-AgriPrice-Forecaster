## Data Types and Missing Values Overview

### Subtask:
Provide a detailed overview of each column's data type, non-null counts, and identify columns with missing values. This will build upon the initial `df.info()` output.


**Reasoning**:
To provide a detailed overview of each column, including data types, non-null counts, missing counts, and their percentages, I will create a new DataFrame that consolidates this information from the existing DataFrame.



```python
total_rows = len(df)
missing_data = []

for col in df.columns:
    data_type = df[col].dtype
    non_null_count = df[col].count()
    missing_count = total_rows - non_null_count
    missing_percentage = (missing_count / total_rows) * 100
    missing_data.append({
        "Column Name": col,
        "Data Type": data_type,
        "Non-Null Count": non_null_count,
        "Missing Count": missing_count,
        "Missing Percentage": f"{missing_percentage:.2f}%"
    })

missing_data_df = pd.DataFrame(missing_data)
print(missing_data_df)
```

**Reasoning**:
The previous step successfully generated a DataFrame with missing value counts and percentages. To fulfill the main task's requirement of displaying missing values ordered from most to least missing, I will sort this `missing_data_df` by the 'Missing Count' column in descending order.



```python
missing_data_df_sorted = missing_data_df.sort_values(by='Missing Count', ascending=False)
print("Missing Values Overview (Sorted by Missing Count):")
print(missing_data_df_sorted)
```

**Reasoning**:
The first part of the main task, detailing missing values, is complete. Now, to fulfill the second part of the main task, I will identify all numerical columns in the DataFrame `df`.



```python
numerical_cols = df.select_dtypes(include=['number']).columns.tolist()
print("Numerical columns identified:", numerical_cols)
```

## Standardize Date Columns

### Subtask:
Convert the 'date' columns in both the `df` and `weather` DataFrames to a consistent datetime format to ensure accurate matching during the merge.


**Reasoning**:
To ensure accurate matching during the merge, I will convert the 'date' columns in both the `df` and `weather` DataFrames to a consistent datetime format using `pd.to_datetime()`.



```python
df['date'] = pd.to_datetime(df['date'])
weather['date'] = pd.to_datetime(weather['date'])

print("Data types after conversion for df:")
print(df['date'].dtype)
print("Data types after conversion for weather:")
print(weather['date'].dtype)
```

**Reasoning**:
The previous step successfully converted the 'date' columns to datetime format. The next instruction in the main task is to rename the 'district' column in the `weather` DataFrame to 'district_name' for proper alignment before merging.



```python
weather = weather.rename(columns={'district': 'district_name'})
print("Columns in weather DataFrame after renaming:")
print(weather.columns.tolist())
```

**Reasoning**:
Now that both DataFrames have consistent 'date' formats and a matching 'district_name' column in the `weather` DataFrame, I will perform a left merge of `df` with `weather` using 'date' and 'district_name' as keys. This will integrate the weather data into the main DataFrame.



```python
df_merged = pd.merge(df, weather, on=['date', 'district_name'], how='left')
df_merged.info()
print("Merged DataFrame shape:", df_merged.shape)
print("First 5 rows of merged DataFrame:")
print(df_merged.head())
```

### Investigating NaN values in weather columns

```python
# Identify rows with NaN values in weather columns
nan_weather_rows = df_merged[df_merged['temp_avg'].isna()]

print(f"Number of rows with NaN in weather columns: {len(nan_weather_rows)}")
print("Sample of rows with missing weather data (date, district_name):")
display(nan_weather_rows[['date', 'district_name', 'temp_avg']].head())
```

```python
# Get unique date-district combinations from the rows with missing weather data
missing_combinations = nan_weather_rows[['date', 'district_name']].drop_duplicates()

print("Unique date-district combinations that resulted in missing weather data:")
display(missing_combinations)

# Check if these combinations exist in the original weather DataFrame
missing_in_weather_df = ~missing_combinations.apply(lambda x: ((weather['date'] == x['date']) & (weather['district_name'] == x['district_name'])).any(), axis=1)

if missing_in_weather_df.any():
    print("\nSome of these date-district combinations are NOT found in the original weather DataFrame:")
    display(missing_combinations[missing_in_weather_df])
else:
    print("\nAll of these date-district combinations are found in the original weather DataFrame. This indicates the issue might be with the join key or data consistency.")
```

## Impute Missing Weather Data

**Reasoning**:
To impute the missing weather data, I will iterate through the specified weather columns, calculate the mean for each, and then use these means to fill the respective NaN values, as per the instructions.



```python
weather_columns = ['temp_avg', 'temp_max', 'temp_min', 'rainfall', 'humidity', 'solar_radiation', 'wind_speed']

print("Imputing missing values in weather columns with their respective means...")

for col in weather_columns:
    mean_value = df_merged[col].mean()
    df_merged[col].fillna(mean_value, inplace=True)
    print(f"Filled NaN in '{col}' with mean: {mean_value:.2f}")

print("\nVerification: Number of NaN values in weather columns after imputation:")
print(df_merged[weather_columns].isnull().sum())
```
